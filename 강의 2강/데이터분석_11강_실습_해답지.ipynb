## Part 0: 환경 설정

```python
# 필요한 라이브러리 임포트
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

# 한글 폰트 설정
plt.rcParams['font.family'] = 'DejaVu Sans'
sns.set_style('whitegrid')

print('필요한 라이브러리를 성공적으로 임포트했습니다!')
```

## Part 1: 데이터 탐색

```python
# 데이터 로드
df = pd.read_csv('customer_data.csv')

# ✓ 데이터의 첫 5행 출력
df.head()
```

```python
# ✓ 데이터셋의 기본 정보 출력
df.info()
```

```python
# ✓ 데이터셋의 통계 요약 출력
df.describe()
```

```python
# ✓ 각 특성의 분포를 시각화하기 (히스토그램)
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
features = ['나이', '월_수입', '구매_경향_점수', '충성도_포인트']
for i, (ax, feat) in enumerate(zip(axes.flat, features)):
    ax.hist(df[feat], bins=30, color='skyblue', edgecolor='black')
    ax.set_title(f'{feat} 분포')
    ax.set_xlabel(feat)
    ax.set_ylabel('빈도')
plt.tight_layout()
plt.show()
```

## Part 2: 데이터 전처리

```python
# 클러스터링에 사용할 특성 선택
features = ['나이', '월_수입', '구매_경향_점수', '충성도_포인트']
X = df[features].copy()

print('선택된 특성:')
print(f'  - 나이 범위: {X["나이"].min()} ~ {X["나이"].max()}')
print(f'  - 월_수입 범위: {X["월_수입"].min()} ~ {X["월_수입"].max()}')
print(f'  - 구매_경향_점수 범위: {X["구매_경향_점수"].min()} ~ {X["구매_경향_점수"].max()}')
print(f'  - 충성도_포인트 범위: {X["충성도_포인트"].min()} ~ {X["충성도_포인트"].max()}')
print('\n⚠️ 범위가 매우 다릅니다. 정규화가 필수입니다!')
```

```python
# ✓ StandardScaler를 사용한 정규화
# 1. StandardScaler 객체 생성
scaler = StandardScaler()

# 2. 데이터 정규화 (fit_transform)
X_scaled = scaler.fit_transform(X)

# 3. 정규화된 데이터를 DataFrame으로 변환
X_scaled_df = pd.DataFrame(X_scaled, columns=features)

# 4. 정규화 후 통계 요약 확인
print('정규화 후 통계:')
print(X_scaled_df.describe())
print('\n✓ 모든 특성의 평균이 0, 표준편차가 1에 가까워졌습니다!')
```

## Part 3: K-means 알고리즘

```python
# ✓ K=3으로 K-means 모델 생성
kmeans_test = KMeans(n_clusters=3, random_state=42, n_init=10)

# ✓ 모델 학습
kmeans_test.fit(X_scaled)

# ✓ 클러스터 할당 결과 확인
print(f'클러스터 중심: \n{kmeans_test.cluster_centers_}')
print(f'\n각 클러스터에 속한 샘플 수:')
cluster_counts = pd.Series(kmeans_test.labels_).value_counts().sort_index()
print(cluster_counts)
```

## Part 4: 최적 클러스터 개수 결정

```python
# ✓ Elbow Method - 각 K에 대한 inertia 계산
inertias = []
K_range = range(1, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertias.append(kmeans.inertia_)

# 그래프 그리기
plt.figure(figsize=(10, 6))
plt.plot(K_range, inertias, 'bo-', linewidth=2, markersize=8)
plt.xlabel('클러스터 개수 (K)', fontsize=12)
plt.ylabel('Inertia', fontsize=12)
plt.title('Elbow Method: 최적 K 찾기', fontsize=14)
plt.grid(True)
plt.show()
print('✓ Elbow 포인트가 보이는 K 값이 최적의 클러스터 개수입니다!')
```

```python
# ✓ Silhouette Score - 각 K에 대한 점수 계산
silhouette_scores = []

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)

# 그래프 그리기
plt.figure(figsize=(10, 6))
plt.plot(K_range, silhouette_scores, 'go-', linewidth=2, markersize=8)
plt.xlabel('클러스터 개수 (K)', fontsize=12)
plt.ylabel('Silhouette Score', fontsize=12)
plt.title('Silhouette Score: 최적 K 찾기', fontsize=14)
plt.grid(True)
plt.show()
print(f'✓ 최고 Silhouette Score: {max(silhouette_scores):.4f} at K={K_range[silhouette_scores.index(max(silhouette_scores))]}')
```

## Part 5: 최적 K로 클러스터링 수행 및 해석

```python
# ✓ 최적 K 설정 (4 권장)
optimal_k = 4

# ✓ 최적 K로 최종 모델 생성 및 학습
kmeans_final = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
clusters = kmeans_final.fit_predict(X_scaled)

# ✓ 원본 데이터프레임에 클러스터 추가
df['클러스터'] = clusters

# ✓ 클러스터별 샘플 수 확인
print(f'클러스터별 샘플 수:')
print(df['클러스터'].value_counts().sort_index())
```

```python
# ✓ 각 클러스터의 특성 분석
print('\n각 클러스터의 평균 특성:')
print(df.groupby('클러스터')[features].mean())
```

```python
# ✓ 2D 시각화 (나이 vs 월_수입)
plt.figure(figsize=(12, 8))

# 각 클러스터별로 다른 색으로 산점도 그리기
colors = ['red', 'blue', 'green', 'orange', 'purple']
for i in range(optimal_k):
    cluster_data = df[df['클러스터'] == i]
    plt.scatter(cluster_data['나이'], cluster_data['월_수입'], 
               label=f'클러스터 {i}', alpha=0.6, s=50, color=colors[i])

plt.xlabel('나이', fontsize=12)
plt.ylabel('월_수입', fontsize=12)
plt.title('고객 클러스터링 결과: 나이 vs 월_수입', fontsize=14)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# ✓ 비즈니스 인사이트 도출
print('\n=== 비즈니스 인사이트 ===')
for i in range(optimal_k):
    cluster_data = df[df['클러스터'] == i]
    print(f'\n클러스터 {i} ({len(cluster_data)}명):')
    print(f'  - 평균 나이: {cluster_data["나이"].mean():.1f}세')
    print(f'  - 평균 월수입: {cluster_data["월_수입"].mean():.0f}만원')
    print(f'  - 평균 구매경향점수: {cluster_data["구매_경향_점수"].mean():.1f}')
    print(f'  - 평균 충성도포인트: {cluster_data["충성도_포인트"].mean():.0f}점')
```

## 도전 문제

### 도전 1: 데이터 정규화의 중요성
정규화 없이 원본 데이터로 K-means를 수행하면 어떻게 될까요?

```python
# ✓ 정규화 없이 K-means 수행
kmeans_unscaled = KMeans(n_clusters=4, random_state=42, n_init=10)
clusters_unscaled = kmeans_unscaled.fit_predict(X)  # 정규화되지 않은 X 사용

# ✓ 결과 비교
print('정규화 전 vs 정규화 후 클러스터 할당 비교:')
print(f'정규화 전 Inertia: {kmeans_unscaled.inertia_:.2f}')
print(f'정규화 후 Inertia: {kmeans_final.inertia_:.2f}')
print('\n✓ 두 결과가 다르다면, 정규화의 중요성을 입증합니다!')

# 더 자세한 비교
print('\n상세 분석:')
print('정규화 전 클러스터 크기:', np.bincount(clusters_unscaled))
print('정규화 후 클러스터 크기:', np.bincount(clusters))
```

### 도전 2: Random State의 영향
다양한 random state로 같은 데이터를 클러스터링하면 결과가 같을까요?

```python
# ✓ 여러 random state로 클러스터링 수행
random_states = [42, 10, 100, 200]
results = []

for rs in random_states:
    km = KMeans(n_clusters=4, random_state=rs, n_init=10)
    labels = km.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    results.append({'Random State': rs, 'Silhouette Score': score})

result_df = pd.DataFrame(results)
print(result_df)
print('\n✓ Silhouette Score는 큰 변화가 없지만, 클러스터 할당은 다를 수 있습니다!')
print('💡 n_init=10을 사용하면 초기화 방법을 여러 번 시도하여 최상의 결과를 선택합니다.')
```

### 도전 3: 다양한 K 값의 시각화
K=2, 3, 4, 5일 때의 클러스터링 결과를 시각적으로 비교하세요.

```python
# ✓ 다양한 K에 대한 클러스터링 시각화
fig, axes = plt.subplots(2, 2, figsize=(14, 12))
K_values = [2, 3, 4, 5]

for idx, k in enumerate(K_values):
    ax = axes.flat[idx]
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    clusters = kmeans.fit_predict(X_scaled)
    
    colors = ['red', 'blue', 'green', 'orange', 'purple']
    for i in range(k):
        mask = clusters == i
        ax.scatter(X_scaled[mask, 0], X_scaled[mask, 1], 
                  label=f'클러스터 {i}', alpha=0.6, color=colors[i])
    
    ax.set_title(f'K={k}일 때의 클러스터링', fontsize=12)
    ax.set_xlabel('정규화된 나이')
    ax.set_ylabel('정규화된 월_수입')
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
print('✓ K가 증가할수록 더 작고 응집된 클러스터들이 생성됩니다.')
```

## 심화 학습: 추가 분석

```python
# 클러스터 프로필 정의
print('\n=== 클러스터 프로필 정의 ===')
profiles = [
    "젊은 저소득층",
    "성인 중산층",
    "중년 고소득층",
    "시니어 고소득층"
]

# 실제 특성 기반으로 프로필 수정
cluster_means = df.groupby('클러스터')[['나이', '월_수입']].mean().round(1)
for i in range(optimal_k):
    age, income = cluster_means.loc[i]
    print(f'\n클러스터 {i}: {profiles[i]}')
    print(f'  평균 나이: {age:.1f}세, 평균 월수입: {income:.0f}만원')
```
